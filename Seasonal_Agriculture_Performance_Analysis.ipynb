```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from scipy import stats

pd.set_option('display.max_columns', None)

df = pd.read_csv("/content/seasonal_agriculture_performance_dataset.csv")

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

df.head()
```

```python
print("Dataset Information:")
df.info()
```

```python
print("Missing Values:")
print(df.isnull().sum())
```

```python
print("Duplicate Rows:")
print(df.duplicated().sum())
```

```python
# Clean missing values using the median

df['Rainfall_mm'] = df['Rainfall_mm'].fillna(df['Rainfall_mm'].median())
df['Soil_Moisture_pct'] = df['Soil_Moisture_pct'].fillna(df['Soil_Moisture_pct'].median())
df['Yield_Tonnes_Ha'] = df['Yield_Tonnes_Ha'].fillna(df['Yield_Tonnes_Ha'].median())

# Check missing values after cleaning
print("Total missing values after cleaning:", df.isnull().sum().sum())

# Check duplicate rows
print("Total duplicate rows:", df.duplicated().sum())
```

```python
# Seasonal Agriculture Performance Analysis

season_summary = df.groupby('Season').agg({
    'Yield_Tonnes_Ha': 'mean',
    'Production_Tonnes': 'mean',
    'Revenue_INR': 'mean',
    'Total_Cost_INR': 'mean',
    'Profit_INR': 'mean'
}).round(2)

season_summary
```

```python
# Average Yield Across Seasons

plt.figure(figsize=(8, 5))

ax = sns.barplot(
    data=df,
    x='Season',
    y='Yield_Tonnes_Ha',
    estimator='mean'
)

plt.title('Average Crop Yield Across Seasons', fontsize=14)
plt.xlabel('Season')
plt.ylabel('Average Yield (Tonnes/Ha)')

# Add values on top of bars
for container in ax.containers:
    ax.bar_label(container, fmt='%.2f')

plt.tight_layout()
plt.show()
```

```python
# Average Profit Across Seasons

profit_summary = df.groupby('Season')['Profit_INR'].mean().round(2)

print("Average Profit by Season:")
print(profit_summary)

plt.figure(figsize=(8, 5))

ax = sns.barplot(
    x=profit_summary.index,
    y=profit_summary.values
)

plt.title('Average Profit Across Seasons', fontsize=14)
plt.xlabel('Season')
plt.ylabel('Average Profit (INR)')

for container in ax.containers:
    ax.bar_label(container, fmt='%.0f')

plt.tight_layout()
plt.show()
```

```python
# Crop-wise Performance Analysis

crop_summary = df.groupby('Crop').agg({
    'Yield_Tonnes_Ha': 'mean',
    'Production_Tonnes': 'mean',
    'Revenue_INR': 'mean',
    'Profit_INR': 'mean'
}).round(2)

# Sort by average profit
crop_summary = crop_summary.sort_values(
    'Profit_INR',
    ascending=False
)

crop_summary
```

```python
# Average Profit by Crop

plt.figure(figsize=(10, 6))

ax = sns.barplot(
    data=crop_summary.reset_index(),
    x='Crop',
    y='Profit_INR'
)

plt.title('Average Profit by Crop', fontsize=14)
plt.xlabel('Crop')
plt.ylabel('Average Profit (INR)')
plt.xticks(rotation=45)

for container in ax.containers:
    ax.bar_label(container, fmt='%.0f', padding=3)

plt.tight_layout()
plt.show()
```

```python
# Best Crop + Season Combination

crop_season_summary = df.groupby(['Crop', 'Season']).agg({
    'Yield_Tonnes_Ha': 'mean',
    'Production_Tonnes': 'mean',
    'Revenue_INR': 'mean',
    'Total_Cost_INR': 'mean',
    'Profit_INR': 'mean'
}).round(2)

# Sort combinations by average profit
crop_season_summary = crop_season_summary.sort_values(
    'Profit_INR',
    ascending=False
)

print("Crop + Season Performance:")
display(crop_season_summary)
```

```python
# Identify the best Crop + Season combination

best_crop = crop_season_summary.index[0][0]
best_season = crop_season_summary.index[0][1]

best_profit = crop_season_summary.iloc[0]['Profit_INR']
best_yield = crop_season_summary.iloc[0]['Yield_Tonnes_Ha']
best_production = crop_season_summary.iloc[0]['Production_Tonnes']
best_revenue = crop_season_summary.iloc[0]['Revenue_INR']
best_cost = crop_season_summary.iloc[0]['Total_Cost_INR']

print("BEST CROP + SEASON COMBINATION")
print("--------------------------------")
print("Crop:", best_crop)
print("Season:", best_season)
print("Average Yield:", best_yield, "Tonnes/Ha")
print("Average Production:", best_production, "Tonnes")
print("Average Revenue: ₹", best_revenue)
print("Average Cost: ₹", best_cost)
print("Average Profit: ₹", best_profit)
```

```python
# Revenue vs Cost vs Profit by Crop

business_summary = df.groupby('Crop')[[
    'Revenue_INR',
    'Total_Cost_INR',
    'Profit_INR'
]].mean().round(2)

business_summary
```

```python
# Plot Revenue, Cost and Profit by Crop

business_summary.plot(
    kind='bar',
    figsize=(12, 6)
)

plt.title('Average Revenue vs Cost vs Profit by Crop', fontsize=14)
plt.xlabel('Crop')
plt.ylabel('Amount (INR)')
plt.xticks(rotation=45)
plt.legend(['Revenue', 'Total Cost', 'Profit'])
plt.tight_layout()
plt.show()
```

```python
# Profit Margin by Crop

profit_margin = (
    df.groupby('Crop')
      .agg(
          Average_Revenue=('Revenue_INR', 'mean'),
          Average_Profit=('Profit_INR', 'mean')
      )
)

profit_margin['Profit_Margin_%'] = (
    profit_margin['Average_Profit'] /
    profit_margin['Average_Revenue']
) * 100

profit_margin = profit_margin.sort_values(
    'Profit_Margin_%',
    ascending=False
).round(2)

profit_margin
```

```python
# Plot Profit Margin by Crop

plt.figure(figsize=(10, 6))

ax = sns.barplot(
    data=profit_margin.reset_index(),
    x='Crop',
    y='Profit_Margin_%'
)

plt.title('Profit Margin by Crop', fontsize=14)
plt.xlabel('Crop')
plt.ylabel('Profit Margin (%)')
plt.xticks(rotation=45)

for container in ax.containers:
    ax.bar_label(container, fmt='%.2f%%', padding=3)

plt.tight_layout()
plt.show()
```

```python
# Most Efficient Crops

efficient_crops = profit_margin[
    ['Average_Revenue', 'Average_Profit', 'Profit_Margin_%']
].copy()

efficient_crops
```

```python
# Rank crops by profitability efficiency

efficient_crops = efficient_crops.sort_values(
    'Profit_Margin_%',
    ascending=False
)

print("Crop Efficiency Ranking:")
print(efficient_crops)
```

# Key Findings

## Key Findings from the Agricultural Performance Analysis

1. **Sugarcane is the most profitable crop**, with an average profit of ₹817,187.99 and the highest profit margin of 59.56%.

5. **The agricultural factor correlations with Sugarcane yield are weak**:
   - Rainfall: 0.283
   - Temperature: 0.093
   - Soil Moisture: 0.171
   - Fertilizer: 0.058
   - Water Usage: 0.074

6. **Rainfall has the strongest relationship with Sugarcane yield among the selected agricultural factors**, although the relationship is still weak and positive.

7. **The correlation heatmap shows that Yield and Production have a strong positive correlation (0.883)**, while Yield and Water Efficiency also show a strong positive correlation (0.913).

8. **Outlier analysis reveals substantial variation in Revenue, Total Cost, Profit, and Water Usage**, indicating that some farms have unusually high or low values.

9. Overall, the analysis shows that **crop choice, season, profitability, and resource efficiency are important dimensions for evaluating agricultural performance**.

# Data-Driven Recommendations

## Recommendations

1. **Prioritize high-profit crops:** Sugarcane and Chilli should be considered strong options because they have the highest average profits and profit margins in the dataset.

2. **Consider seasonal performance:** Kharif Sugarcane should receive particular attention because Sugarcane + Kharif is the best-performing Crop + Season combination identified in the analysis.

3. **Review low-profit crops:** Wheat, Rice, Maize, and Pulses show negative average profits. Their production costs, market prices, and yield performance should be reviewed before expanding their cultivation.

4. **Improve resource efficiency:** Since Water Usage has a very weak correlation with Sugarcane yield (0.074), increasing water usage alone may not result in higher yield. Efficient water management should therefore be emphasized.

5. **Optimize fertilizer usage:** The fertilizer-yield correlation for Sugarcane is very weak (0.058). Fertilizer application should therefore be optimized rather than increased without considering other factors.

6. **Monitor environmental conditions:** Rainfall has the strongest correlation among the selected agricultural factors with Sugarcane yield (0.283). Rainfall and soil moisture should be monitored as part of crop planning.

7. **Investigate outliers:** Farms with unusually high or low revenue, cost, profit, or water usage should be examined to determine whether they represent exceptional performance, large-scale operations, or unusual conditions.

8. **Use multiple factors for decision-making:** Since the individual environmental factors show weak correlations with Sugarcane yield, agricultural decisions should not be based on a single factor. Crop, season, resources, production cost, revenue, and market price should be considered together.

9. **Focus on profitability as well as yield:** A high yield does not necessarily guarantee high profit. Both production performance and financial performance should be considered when selecting crops and cultivation strategies.

# Limitations

## Limitations of the Analysis

1. The analysis is based only on the variables available in the provided dataset and may not capture all real-world agricultural factors.

2. Correlation analysis measures linear association between variables and does not establish causation.

3. Missing numerical values were replaced using median imputation, which may not completely represent the original observations.

4. The presence of outliers may influence some averages and statistical relationships.

5. The dataset represents recorded observations and therefore the findings should not automatically be generalized to all farms, regions, crops, or seasons.

6. The analysis identifies patterns and relationships in the available data, but further field-level or longitudinal data would be required to validate the findings.
